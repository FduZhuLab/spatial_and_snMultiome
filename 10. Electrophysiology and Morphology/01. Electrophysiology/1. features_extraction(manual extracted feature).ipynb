```python
import numpy as np
import pandas as pd
import os
from scipy import optimize as op
from itertools import groupby
from scipy import signal
import math
from pathlib import Path
from operator import itemgetter
from itertools import groupby
from matplotlib import pyplot as plt
import shutil
import sys
sys.path.insert(0, r'C:\Users\nVoke\PycharmProjects\pythonProject1\tools')
from scan import scan_file
from mutiple_group_test import my_oneWayAnova

%matplotlib inline
sample_rate = 20000
duration=200
```

```python
root_dir=r'F:\jcy\humen_brain\electrophysiology\all_nerons\AP\all'
newPath=r'F:\jcy\humen_brain\electrophysiology\all_nerons\AP\inhibitory'
root=Path(root_dir)
file_info=[]

file_info = scan_file(root, file_info)
df = pd.DataFrame(file_info, columns=['filename', 'ext', 'dir'])
para=[]
n=0
for fname, dir in zip(df.filename, df.dir):
    if fname.endswith('AP.txt'):
        valid_peak=[]
        fpath = os.path.join(dir, fname)
        file=pd.read_csv(fpath, sep='\t')
        x = file['trace'].values
        time_range=0.03
        peak_distance = sample_rate * 0.1
        peaks, _ = signal.find_peaks(x, height=-20, distance=peak_distance,prominence=1)

        peaks=[peak for peak in peaks if len(x)-duration>peak>duration]
        peak_cell,slop_cell,length = [],[],[]
        thre_all,amp_all,half_width_all,DepoR_all,RepoR_all,AHPAmp_all=[],[],[],[],[],[]
        for p in peaks:
            peak_around=x[p-duration:p+duration]
            invalid=[x[0] for x in list(enumerate(peak_around)) if x[1]>-20]
            ranges = []
            for k, g in groupby(enumerate(invalid),lambda x:x[0]-x[1]):
                group=list(map(itemgetter(1), g))
                ranges.append((group[0], group[-1]))
            if len(ranges)==1:
                valid_peak.append(p)
        if len(valid_peak)==0:
            continue

        for p in valid_peak:
            peak_p=x[p-duration:p+duration]
            slope_p=np.diff(x[p-duration:p+duration+1])
            thre_p=peak_p[np.where(slope_p >0.1*max(slope_p))][0]
            thre_all.append(thre_p)
            amp_p=max(peak_p)-thre_p
            amp_all.append(amp_p)
            half_width_p=len(np.where(peak_p>0.5*max(peak_p))[0])/20
            half_width_all.append(half_width_p)
            DepoR_p=max(slope_p)
            DepoR_all.append(DepoR_p)
            RepoR_p=min(slope_p)
            RepoR_all.append(RepoR_p)
            AHPAmp_p=thre_p-min(peak_p[duration:])
            AHPAmp_all.append(AHPAmp_p)

            peak_cell.append(peak_p)
            length.append(len(peak_p))

            slop_cell.append(slope_p)
        peak_i =np.mean(peak_cell,axis=0).reshape(-1)
        slop_i =np.mean(slop_cell,axis=0).reshape(-1)*20

        thre=np.mean(thre_all)
        amp=np.mean(amp_all)
        half_width=np.mean(half_width_all)
        DepoR=np.mean(DepoR_all)
        RepoR=np.mean(RepoR_all)
        AHPAmp=np.mean(AHPAmp_all)
        para.append([fname,thre,amp,half_width,DepoR,RepoR,AHPAmp])
        plt.figure(figsize=(9,3))
        grid=plt.GridSpec(1,5) ###分别设置子图大小

        plt.subplot(grid[0,:2]);plt.plot(peak_i,label=f'{fname},\n,half_wid:{half_width},\n,AHPamp:{np.round(AHPAmp,2)}')
        plt.legend(loc=1)
        plt.subplot(grid[0,2:]);plt.plot(x,c='r');plt.scatter(valid_peak,x[valid_peak],c='g')
        plt.tight_layout()
        plt.show()

        # if half_width<2.5 and AHPAmp>8:
        #     n+=1
        #     # print(f'编号：{n}, half_width:{half_width}, {AHPAmp}')
        #     newDir=os.path.join(newPath,fname)
        #     shutil.move(fpath,newDir)

pra=pd.DataFrame(para)
pra.columns=['cell_name','peak_thre','peak_amp','half_width','DepoR','RepoR','AHPAmp']
# pra.set_index('cell_name',drop=True)
print(pra.head())

plt.subplot(211);pra['half_width'].hist(bins=50)
plt.subplot(212);pra['AHPAmp'].hist(bins=50)
# plt.tight_layout()
plt.show()
```
